```python
!pip install matplotlib
import pandas as pd
import matplotlib.pyplot as plt


# Load cleaned  GDP data
file_path = "Filtered_African_Countries_GDP_Data__2020_Onwards_.csv"  # Ensure this file exists
df = pd.read_csv(file_path)

# Convert numeric columns to float (handling 'no data' or non-numeric values)
df = df.replace("no data", None)
df.iloc[:, 1:] = df.iloc[:, 1:].astype(float)

# Set "Country" as the index
df.set_index('Countries', inplace=True)

###LINE CHART: GDP Trends Over Time
plt.figure(figsize=(14, 8))
for country in df.index:
    plt.plot(df.columns, df.loc[country], marker='', label=country)

plt.xlabel("Year", fontsize=28)
plt.ylabel("GDP per Capita (USD)" , fontsize=28)
plt.title("GDP per Capita Trends for All African Countries" , fontsize=30)
plt.legend(loc="upper left", bbox_to_anchor=(1, 1), fontsize="20", ncol=2)
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```
